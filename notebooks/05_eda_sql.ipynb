```python
import sqlite3
import pandas as pd

# Create SQLite database connection
conn = sqlite3.connect('spacex.db')
cursor = conn.cursor()

# Load CSV to SQL table
df = pd.read_csv("Spacex.csv")
df.to_sql("SPACEXTBL", conn, if_exists='replace', index=False)

# Sample Queries required for Capstone
query_1 = "SELECT DISTINCT Launch_Site FROM SPACEXTBL;"
print("Launch Sites:")
print(pd.read_sql(query_1, conn))

query_2 = "SELECT SUM(PAYLOAD_MASS__KG_) FROM SPACEXTBL WHERE Customer = 'NASA (CRS)';"
print("\nTotal Payload Mass by NASA (CRS):")
print(pd.read_sql(query_2, conn))

conn.close()
```
